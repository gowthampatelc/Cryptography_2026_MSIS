```python
with open("cryptology.txt", "r") as file:
    lines = file.readlines()

    num_lines =  len(lines)
    num_words = 0
    num_letters = 0

    for i in lines:
        num_words+=len(i.split())
        num_letters += len(i)

print("Number of lines:",num_lines)
print("Number of words:", num_words)
print("Number of letters:", num_letters)
```

```python
with open("cryptology_no.txt", "r") as file_no:
    lines = file_no.read()

    upc = 0
    lpc = 0
    dig = 0
    spa = 0
    spchr = 0

    for i in (lines):
        if i.isupper():
            upc+=1
        elif i.islower():
            lpc+=1
        elif i.isdigit():
            dig+=1
        elif i == " ":
            spa+=1
        else:
            spchr+=1

print("Upper case letters :", upc)
print("Lower case letters :", lpc)
print("Digits :", dig)
print("Space :", spa)
print("Special characters :", spchr)
    
```

```python
with open("input.txt", "w") as file:
    for i in range(5):
        text = input("Enter line " + str(i + 1) + ": ")
        file.write(text + "\n")

with open("input.txt", "r") as file:
    content = file.read()

lines = len(content.splitlines())
words = len(content.split())
alphabets = sum(c.isalpha() for c in content)
digits = sum(c.isdigit() for c in content)
spaces = sum(c.isspace() for c in content)
special = sum(not c.isalnum() and not c.isspace() for c in content)

print("Number of lines:", lines)
print("Number of words:", words)
print("Alphabetic characters:", alphabets)
print("Digits:", digits)
print("Spaces:", spaces)
print("Special characters:", special)

with open("processed.txt", "w") as file:
    file.write(content.upper())

with open("processed.txt", "r") as file:
    print("\nProcessed contents:")
    print(file.read())
```

```python
import re

with open("security_data.txt", "r") as file:
    file_content = file.read()

ip_addresses = re.findall(r'\b(?:\d{1,3}\.){3}\d{1,3}\b', file_content)
emails = re.findall(r'\b[\w.-]+@[\w.-]+\.\w+\b', file_content)
session_ids = re.findall(r'\bSEC-\d{4}-\d{4}\b', file_content)
hex_values = re.findall(r'\b(?=[0-9a-fA-F]*[a-fA-F])[0-9a-fA-F]+\b', file_content)

with open("pattern_results.txt", "w") as file:
    file.write("IP Addresses:\n")
    for value in ip_addresses:
        file.write(value + "\n")

    file.write("\nEmail Addresses:\n")
    for value in emails:
        file.write(value + "\n")

    file.write("\nSession IDs:\n")
    for value in session_ids:
        file.write(value + "\n")

    file.write("\nHexadecimal Sequences:\n")
    for value in hex_values:
        file.write(value + "\n")

print("Patterns stored in pattern_results.txt")
```
