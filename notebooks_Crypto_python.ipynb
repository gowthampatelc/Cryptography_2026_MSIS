```python
text = "CRYPTOLOGYISINTERESTING"

frequencies = {}

for char in text:
    if char in frequencies:
        frequencies[char] += 1
    else:
        frequencies[char] = 1

print("Character Frequencies:")
for char, count in frequencies.items():
    print(f"{char}: {count}")

#Find the highest count
max_count = 0
for count in frequencies.values():
    if count > max_count:
        max_count = count

#Identify all characters that have that highest count
most_frequent = []
for char, count in frequencies.items():
    if count == max_count:
        most_frequent.append(char)

print(f"\nMost frequent characters: {', '.join(most_frequent)} ({max_count} times)")
```

```python
text = "CRYPTOLOGYISINTERESTING"

frequencies = {}
for char in text:
    if char in frequencies:
        frequencies[char] += 1
    else:
        frequencies[char] = 1

#Get the total number of characters
total_chars = len(text)

print(f"Total Characters: {total_chars}\n")
print("Character Frequencies and Percentages:")
print(f"{'Char':<5}  {'Count':<5}  {'Percentage'}")
print("-" * 25)

#Calculate and display percentage frequency
for char, count in frequencies.items():
    percentage = (count / total_chars) * 100
    print(f"{char:<5}  {count:<5}  {percentage:.2f}%")
```

```python
def check_prime(n):
    if n <= 1:
        return False
    
    for i in range(2, n):
        if n % i == 0:
            return False 
            
    return True 

num = int(input("Enter a number to check: "))
if check_prime(num):
    print(num, "is a prime number.\n")
else:
    print(num, "is NOT a prime number.\n")

start = int(input("Enter the start of the range: "))
end = int(input("Enter the end of the range: "))

print("Prime numbers in this range:")
for n in range(start, end + 1):
    if check_prime(n):
        print(n, end=" ")
```

```python
# Read two integers from the user
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

a = num1
b = num2

while b != 0:
    remainder = a % b
    a = b
    b = remainder

gcd = a

print(f"\nThe GCD of {num1} and {num2} is {gcd}.")

if gcd == 1:
    print(f"{num1} and {num2} are coprime.")
else:
    print(f"{num1} and {num2} are NOT coprime.")
```

```python
b = int(input("Enter the integer (b): "))
m = int(input("Enter the modulo (m): "))

inverse = None

for x in range(1, m):
    if (b * x) % m == 1:
        inverse = x
        break 

if inverse != None:
    print(f"\nThe modular multiplicative inverse of {b} mod {m} is {inverse}.")
else:
    print(f"\nNo inverse exists because {b} and {m} are not coprime.")
```
