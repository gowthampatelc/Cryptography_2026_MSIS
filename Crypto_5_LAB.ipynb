```python
def shift_encrypt(text, shift):
    result = []
    for char in text:
        if char.isalpha():
            ascii_conv = ord('A') if char.isupper() else ord('a')
            result.append(chr((ord(char) - ascii_conv + shift) % 26 + ascii_conv))
        else:
            result.append(char)
    return "".join(result)

def shift_decrypt(text, shift):
    return shift_encrypt(text, -shift)


message = input("Enter word to be encrypted using sgift cipher ")
key = int(input("Enter the shift "))
encrypted = shift_encrypt(message, key)
decrypted = shift_decrypt(encrypted, key)

print("Shift Encrypted:", encrypted)
print("Shift Decrypted:", decrypted)
```

```python
def rot13(text):
    result = []
    for char in text:
        if char.isalpha():
            ascii_conv = ord('A') if char.isupper() else ord('a')
            result.append(chr((ord(char) - ascii_conv + 13) % 26 + ascii_conv))
        else:
            result.append(char)
    return "".join(result)

message = input("Enter word to be encrypted using ROT13 ")
encrypted = rot13(message)
decrypted = rot13(encrypted)

print("ROT13 Encrypted:", encrypted)
print("ROT13 Decrypted:", decrypted)
```

```python
def atbash(text):
    result = []
    for char in text:
        if char.isupper():
            result.append(chr(ord('Z') - (ord(char) - ord('A'))))
        elif char.islower():
            result.append(chr(ord('z') - (ord(char) - ord('a'))))
        else:
            result.append(char)
    return "".join(result)

message = input("Enter word to be encrypted using atbash cipher ")
encrypted = atbash(message)
decrypted = atbash(encrypted)

print("Atbash Encrypted:", encrypted)
print("Atbash Decrypted:", decrypted)
```

```python
def _mod_inverse(a, m):
    for x in range(1, m):
        if (a * x) % m == 1:
            return x
    raise ValueError(f"Modular inverse does not exist for a={a} and m={m}")

def affine_encrypt(text, a, b):
    result = []
    for char in text:
        if char.isalpha():
            ascii_conv = ord('A') if char.isupper() else ord('a')
            x = ord(char) - ascii_conv
            encrypted = (a * x + b) % 26
            result.append(chr(encrypted + ascii_conv))
        else:
            result.append(char)
    return "".join(result)

def affine_decrypt(text, a, b):
    result = []
    a_inv = _mod_inverse(a, 26)
    for char in text:
        if char.isalpha():
            ascii_conv = ord('A') if char.isupper() else ord('a')
            y = ord(char) - ascii_conv
            decrypted = (a_inv * (y - b)) % 26
            result.append(chr(decrypted + ascii_conv))
        else:
            result.append(char)
    return "".join(result)

# --- Example Usage ---
message = input("Enter word to be encrypted using affine cipher ")
a_key, b_key = 5, 8  # 'a' must be coprime to 26
encrypted = affine_encrypt(message, a_key, b_key)
decrypted = affine_decrypt(encrypted, a_key, b_key)

print("Affine Encrypted:", encrypted)
print("Affine Decrypted:", decrypted)
```
