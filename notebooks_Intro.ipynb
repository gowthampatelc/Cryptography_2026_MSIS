```python
message = "CRYPTOLOGY USING PYTHON"
print(len (message))
for i in range(0,5):
    print(message[i],end="")
print()
for i in message[-6:]:
    print(i , end="")
print()
for i in message[3:9]:
    print(i , end="")
print()
for i in message[::-1]:
    print(i , end="")
print()
for i in message[::]:
    print(i,end="")
```
