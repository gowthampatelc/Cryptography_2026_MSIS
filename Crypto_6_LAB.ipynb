```python
def otp_encrypt(plaintext, key):
    if len(key) < len(plaintext):
        raise ValueError("Key length must be at least as long as the plaintext.")
    
    result = []
    for p, k in zip(plaintext, key):
        if p.isalpha():
            offset = ord('A') if p.isupper() else ord('a')
            k_offset = ord('A') if k.isupper() else ord('a')
            
            p_val = ord(p) - offset
            k_val = ord(k) - k_offset
            encrypted_val = (p_val + k_val) % 26
            result.append(chr(encrypted_val + offset))
        else:
            result.append(p)
    return "".join(result)

def otp_decrypt(ciphertext, key):
    if len(key) < len(ciphertext):
        raise ValueError("Key length must be at least as long as the ciphertext.")
    
    result = []
    for c, k in zip(ciphertext, key):
        if c.isalpha():
            offset = ord('A') if c.isupper() else ord('a')
            k_offset = ord('A') if k.isupper() else ord('a')
            
            c_val = ord(c) - offset
            k_val = ord(k) - k_offset
            decrypted_val = (c_val - k_val) % 26
            result.append(chr(decrypted_val + offset))
        else:
            result.append(c)
    return "".join(result)

message = input("Enter word to be encrypted using OTP ")
otp_key = input("Enter OTP key ")

encrypted_otp = otp_encrypt(message, otp_key)
decrypted_otp = otp_decrypt(encrypted_otp, otp_key)

print("OTP Encrypted:", encrypted_otp)
print("OTP Decrypted:", decrypted_otp)
```

```python
def brute_force_shift(ciphertext):
    print(f"Brute-forcing ciphertext: '{ciphertext}'\n")
    for shift in range(26):
        result = []
        for char in ciphertext:
            if char.isalpha():
                ascii_offset = ord('A') if char.isupper() else ord('a')
                # Decrypting by shifting backwards
                result.append(chr((ord(char) - ascii_offset - shift) % 26 + ascii_offset))
            else:
                result.append(char)
        print(f"Shift {shift:2d}: {''.join(result)}")

cipher_text = "MCI RWR WH"
brute_force_shift(cipher_text)
```

```python
def rail_fence_encrypt(text, rails):
    if rails <= 1:
        return text
    
    # Create a list of empty lists for each rail using a complete for loop
    fence = []
    for _ in range(rails):
        fence.append([])
        
    rail = 0
    direction = 1  # 1 means moving down, -1 means moving up
    
    for char in text:
        fence[rail].append(char)
        rail += direction
        # Reverse direction when hitting the top or bottom rail
        if rail == 0 or rail == rails - 1:
            direction *= -1
            
    # Flatten the fence to produce the final ciphertext using a complete for loop
    ciphertext = ""
    for r in fence:
        for char in r:
            ciphertext += char
            
    return ciphertext

message = input("Enter word to be encrypted using rail fence cipher ")
num_rails = int(input("Enter the depth of the rails"))

encrypted_rail = rail_fence_encrypt(message, num_rails)
print("Rail Fence Encrypted:", encrypted_rail)
```
